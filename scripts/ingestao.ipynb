## CAMADA BRONZE INGESTÃO DE DADOS

```python
import duckdb
import pandas as pd
import os
from datetime import datetime
```

```python
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
#arquivo = 'z0019_1.csv'
arquivo = 'z0019_2.csv'
data_ingestao = datetime.now()
#criação de dataframe com pandas
df = pd.read_csv(f'../landing/{arquivo}', sep=';')

#criação de colunas no dataframe para visualização
df['nome_arquivo'] = arquivo
df['data_ingestao'] = data_ingestao
df.head()
```

```python
con.execute("""
    CREATE TABLE IF NOT EXISTS bronze_z0019 (
        NATBR VARCHAR,
        MAKTX VARCHAR,
        WERKS VARCHAR,
        MAINS VARCHAR,
        LABST VARCHAR,
        nome_arquivo VARCHAR,
        data_ingestao TIMESTAMP
    )
""")
```

```python
con.execute("INSERT INTO bronze_z0019 SELECT * FROM df")
```

```python
resultado = con.execute("SELECT * FROM bronze_z0019").fetch_df()
resultado.head(6)
```

```python
con.close()
```
